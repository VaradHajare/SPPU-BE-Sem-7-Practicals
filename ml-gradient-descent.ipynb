```python
'''Implement Gradient Descent Algorithm to find the local minima of a function. 
For example, find the local minima of the function y=(x+3)² starting from the point x=2.''' 
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Function
x_values = np.linspace(-6, 4, 100)
y_values = (x_values + 3) ** 2

# Gradient Descent points
x = 2
lr = 0.1
points = [x]

for i in range(20):
    x = x - lr * 2 * (x + 3)
    points.append(x)

# Plot
plt.plot(x_values, y_values, label="y = (x+3)^2")
plt.scatter(points, [(p+3)**2 for p in points],
            color="red", label="Gradient Descent")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent Optimization")
plt.legend()
plt.grid()
plt.show()
```

```python
# Define the function
def f(x):
    return (x + 3) ** 2

# Define derivative
def df(x):
    return 2 * (x + 3)

# Initialize parameters
x = 2
learning_rate = 0.1
iterations = 100

# Gradient Descent
for i in range(iterations):
    gradient = df(x)
    x = x - learning_rate * gradient

    print("Iteration:", i + 1, "x =", x, "y =", f(x))

# Final result
print("\nMinimum x:", x)
print("Minimum y:", f(x))
```
